```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame(spark.sql("SELECT * FROM Schema.table_name_Ram").toPandas())
df.head()
```

```python
import matplotlib.pyplot as plt
total_back_sales = df.groupby('CustomerName')['Order$'].sum()
total_back_sales
```

```python
# plt.rcParams['figure.figsize'] = [10, 5] or you can use below line
plt.figure(figsize=(20,10))
plt.plot(total_back_sales)
plt.title('far by analysis')
plt.xlabel('Customer Name')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
# bar graphs
# histograms
# pie charts
# scater plots
```

```python
#Bar chart
plt.figure(figsize=(10,5))
sales_by_region = df.groupby('Region')['Order$'].sum()
plt.bar(sales_by_region.index, sales_by_region.values)
plt.title('Sales by Region')
plt.xticks(rotation=45)
# plt.plot(kind = 'bar',color = 'red')
```

```python
#Histogram
plt.figure(figsize=(10,5))
plt.hist(df.groupby('SalesRep Name')['Order$'].sum(), bins=20 , color = 'Red' , edgecolor = 'Yellow')
plt.title('sales by sales rep')
plt.xlabel('OrderQty')
```

```python
plt.rcParams['figure.figsize'] = [10,5]
Selling_price_unique = df['Unit Selling Price'].value_counts()
# plt.pie(df.groupby('Unit Selling Price')['Order$'].sum(), labels=Selling_price_unique.index, autopct='%1.1f%%')
plt.pie(Selling_price_unique, labels=Selling_price_unique.index, autopct='%1.1f%%',startangle=90)
plt.title('Sales by Selling Price')
plt.show()
```
