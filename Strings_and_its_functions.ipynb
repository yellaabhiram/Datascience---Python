6. **string** **formatting**

```python
"Hello my name is {} and I am {} years old ".format("sau", 29)
```

```python
"Hello my name is {1} and I am {0} years old".format("sau", 29)
```

```python
"Hello my name is {name} and I am {age} years old".format(name = "sau", age= 29)
```

```python
"Hello my name is {name} and I am {age} years old and I am a {preofession}".format(name = "sau", age= 29, preofession = "AI engi")
```

```python
# name %s is a placeholder for strings, %d is the placeholder for integers
name = "sau"
age = 28
formatted_string = "Hello my name is %s and I am %d years old"%(name, age)
print(formatted_string)
```

```python
# f strings
name = "sau"
age = 28
formatted_string = f"Hello my name is {name} and I am {age} years old"
print(formatted_string)
```
