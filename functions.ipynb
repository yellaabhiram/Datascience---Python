Arbitrary Keyword Arguments, **kwargs


```python
def calculate_factorial(number):
  factorial = 1
  for i in range(1,number+1):
    factorial *= i
  return factorial
```

```python
#  create function for combination

def calculate_combination(n,r):
  n_factorial = calculate_factorial(n)
  r_factorial = calculate_factorial(r)
  n_minus_r_factorial = calculate_factorial(n-r)
  combination = n_factorial / (r_factorial * n_minus_r_factorial)
  return combination
```

```python
calculate_combination(5,2)
```
