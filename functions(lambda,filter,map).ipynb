**filter() in python**

```python
def fun(variable):
  letters = ['a', 'e', 'i', 'o', 'u']
  if (variable in letters):
    return True
  else:
    return False

sequence = ['g', 'e', 'e', 'j', 'k', 's', 'p', 'r', 'a', 'o','v','k','i',]

filtered = filter(fun, sequence)

print('The filtered letters are:')
for s in filtered:
  print(s)
```

```python
# ususally filtwr functions are used with lambda functions

# what is lambda function
# Python Lambda Functions are anonymous functions means that the function is without a name.

#Syntax: lambda arguments : expression

l1 = [1,2,3,4,6]
l2 = [19,5,9,8,7]



mutuplyby7 = lambda x: x*7
for i in l1:
  print(mutuplyby7(i))
for i in l2:
  print(mutuplyby7(i))


```

```python
is_even_list = [lambda arg=x: arg * 10 for x in range(1, 9)]
for item in is_even_list:
  print(item())
# iterate on each lambda function
# and invoke the function to get the calculated value
```

```python
# lambda function with if else

max = lambda a, b : a if(a > b) else b
print(max(1, 0.5))
```

```python
# Python Lambda with Multiple Statements
List = [[2,3,4],[1, 4, 16, 64],[3, 6, 9, 12]]

sortlist = lambda x : (sorted(i) for i in x)
# print(sortlist(List))
# print(sortlist)

secondlargest = lambda x, f : [y[-2] for y in f(x)]
res = secondlargest(List, sortlist)
print(res)


#try to solve this question without using lambda function , find the second largest element from each sublist present in the list!
```

```python
# Using lambda() Function with filter()
# filter takes list as a argument input, function list
# Filter out all odd numbers using filter() and lambda function
l = [5,3,4,6,7,9,11,23,12,3,4,67,78,98]   # do the same queation without using filter and lambda

final_list = list(filter(lambda x : (x%2 !=0 ), l))
final_list
```

```python
ages = [13, 90, 17, 59, 21, 60, 5]
adults  =list(filter(lambda age: age>18, ages))
adults
```
