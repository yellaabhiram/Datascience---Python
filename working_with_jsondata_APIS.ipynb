```python
def basic_json():

  print("\n==== Example 1 : Basic of json Opeatations ====")

  # create a json object (dict in python)

  student = {
      "name": "Alice Smith",
      "age": 20,
      "grades": [85, 92, 88],
      "contact": {
          "email": "alice@example.com",
          "phone": "123-456-7890"
      },
      "active": True
  }

  # covert this data dict into json

  json_string = json.dumps(student, indent = 4)
  print(json_string )


      # Converting JSON string back to Python dictionary
  parsed_data = json.loads(json_string)
  print("\nAccessing specific values:")
  print(f"Name: {parsed_data['name']}")
  print(f"First grade: {parsed_data['grades'][0]}")
  print(f"Email: {parsed_data['contact']['email']}")
```

```python
basic_json()
```

```python
# file opertions !

def file_operations():

  """
  reading and writing json files
  """
  print("\n==== Example 2 : working with files ====")

    # Sample data
  class_data = {
      "class_name": "Python Programming",
      "students": [
          {"id": 1, "name": "John", "score": 95},
          {"id": 2, "name": "Emma", "score": 88},
          {"id": 3, "name": "Michael", "score": 92}
      ]
  }

  # write the file in json

  with open("class_data.json", "w") as file:
    json.dump(class_data, file, indent = 4)

  print("\nJSON data written to 'class_data.json'")

  # reading the json file

  with open("class_data.json", "r") as file:
    data = json.load(file)

  print("\nJSON data read from 'class_data.json':")

  for student in data["students"]:
    print(f"Name: {student['name']}, Score: {student['score']}")
```

```python
file_operations()
```

```python
import requests
import json
from requests.exceptions import RequestException, Timeout, HTTPError

def example_without_error_handling():
    """
    Example 1: API request without error handling
    This approach is risky and might crash with various errors
    """
    print("\n=== Example 1: Without Error Handling ===")

    # Getting user data from JSONPlaceholder API
    url = "https://jsonplaceholder.typicode.com/users/1"

    # Making the request without any error handling
    response = requests.get(url)
    user_data = response.json()

    # Processing the response
    print(f"User Name: {user_data['name']}")
    print(f"User Email: {user_data['email']}")
    print(f"User Company: {user_data['company']['name']}")


def example_with_error_handling():
    """
    Example 2: API request with comprehensive error handling
    This approach safely handles various types of errors that might occur
    """
    print("\n=== Example 2: With Error Handling ===")

    # Same API endpoint
    url = "https://jsonplaceholder.typicode.com/users/1"

    try:
        # Setting timeout and headers for better reliability
        headers = {
            'User-Agent': 'Python-Tutorial-App',
            'Accept': 'application/json'
        }

        # Making the request with timeout
        response = requests.get(url, headers=headers, timeout=5)

        # Check if the request was successful
        response.raise_for_status()

        try:
            # Try to parse JSON response
            user_data = response.json()

            # Verify expected data exists before accessing
            if all(key in user_data for key in ['name', 'email', 'company']):
                print(f"User Name: {user_data['name']}")
                print(f"User Email: {user_data['email']}")
                print(f"User Company: {user_data['company']['name']}")
            else:
                print("Response is missing expected data fields")

        except json.JSONDecodeError as e:
            print(f"Error decoding JSON response: {e}")
            print(f"Raw response: {response.text[:100]}")  # Show first 100 chars

    except Timeout:
        print("Request timed out - server took too long to respond")

    except HTTPError as e:
        if e.response.status_code == 404:
            print("User not found")
        elif e.response.status_code == 401:
            print("Authentication required")
        elif e.response.status_code == 403:
            print("Access forbidden")
        else:
            print(f"HTTP Error occurred: {e}")

    except RequestException as e:
        print(f"Network error occurred: {e}")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")


def demonstrate_error_cases():
    """
    Example 3: Demonstrating different error scenarios
    """
    print("\n=== Example 3: Error Scenarios ===")

    # 1. Invalid URL (will cause connection error)
    try:
        response = requests.get("https://nonexistent.example.com")
    except RequestException as e:
        print(f"\nConnection Error Demo:\n{e}")

    # 2. Invalid endpoint (will cause 404)
    try:
        response = requests.get("https://jsonplaceholder.typicode.com/nonexistent")
        response.raise_for_status()
    except HTTPError as e:
        print(f"\n404 Error Demo:\n{e}")

    # 3. Invalid JSON response
    try:
        invalid_json = "{ this is not valid JSON }"
        data = json.loads(invalid_json)
    except json.JSONDecodeError as e:
        print(f"\nJSON Decode Error Demo:\n{e}")


if __name__ == "__main__":
    print("\nRunning example without error handling (might crash)...")
    try:
        example_without_error_handling()
    except Exception as e:
        print(f"Example 1 crashed with error: {e}")

    print("\nRunning example with error handling (handles errors gracefully)...")
    example_with_error_handling()

    print("\nDemonstrating various error scenarios...")
    demonstrate_error_cases()
```
