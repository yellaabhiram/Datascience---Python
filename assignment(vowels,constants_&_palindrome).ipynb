```python

def function(string,vowels):
  result = [one for one in string if one in vowels]
  result2 = [one for one in string if one in constants]
  print(len(result))
  print(len(result2))

vowels = 'aeiou'
constants = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
string = input("give your string: ")
function(string,vowels)
```

```python
def palindorme_fun(string):
  if string == string[::-1]:
    return True
  else:
    return False

string = input("give your string: ")
palindorme_fun(string)
```
