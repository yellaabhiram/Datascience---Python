```python
def convert_to_list(str):
  return eval(str)
```

```python
# we want to apply this function to non null values
df.loc[~df["genres"].isna(), "genres"] = df.loc[~df["genres"].isna(), "genres"].apply(convert_to_list)
```

```python
df.loc[~df["cast"].isna(), "cast"] = df.loc[~df["cast"].isna(), "cast"].apply(convert_to_list)
```

```python
df.loc[~df["crew"].isna(), "crew"] = df.loc[~df["crew"].isna(), "crew"].apply(convert_to_list)
```

```python
df_new["profit"] = df_new["revenue"] - df_new["budget"]
df_new["roi"] = 100* (df_new["profit"] / df_new["budget"])
```

```python
max_profit_movie_df['cast'][0]["name"]
```

```python
crew_list = max_profit_movie_df.loc['crew']
crew_list[0:3]
```

```python
producer_list = []
director_list = []

for elem in crew_list:
  if elem["job"] == "Producer":
    producer_list.append(elem["name"])
  if elem["job"] == "Director":
    director_list.append(elem["name"])
```

```python
print(f"Producer: {producer_list}")
print(f"Director: {director_list}")
```

```python
cast_list = max_profit_movie_df['cast']
cast_list[0:3]
```

```python
#actors in the highest profit movie
actor_list = []
for elem in cast_list:
  actor_list.append(elem["name"])
```

```python
print("Actors in the highest profit movie: ")
print(actor_list[0:5])
```

2. **Find out the unique genres of movies in this dataset.**

```python
genres_without_nullvalues= df.loc[~df["genres"].isna()]
```

```python
len(genres_without_nullvalues)
```

```python
genres_without_nullvalues.loc[5,'genres']
```

```python
genres_list = []
for index,row in genres_without_nullvalues.iterrows():
  generes = genres_without_nullvalues.loc[index,"genres"]
  for k in generes:
    genres_list.append(k["name"])

pd.DataFrame(set(genres_list),columns=["Unique_Genres"])
```
